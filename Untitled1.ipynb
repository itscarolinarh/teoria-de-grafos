## Introducción
Este tutorial ilustra la implementación, formulación y resolución de tres problemas fundamentales de optimización sobre redes mediante la biblioteca `NetworkX` de Python[cite: 1, 2]:
1. **Árbol de Expansión Mínima (Minimum Spanning Tree - MST)**
2. **Ruta más corta (Shortest Path)**
3. **Flujo Máximo (Maximum Flow)**

Las redes se representan matemáticamente mediante un grafo $G = (V, E)$, donde $V$ es el conjunto de vértices o nodos y $E$ es el conjunto de aristas o arcos.

```python
# Importación de librerías requeridas
import networkx as nx
import matplotlib.pyplot as plt

# Ajustamos el tamaño predeterminado de las figuras
plt.rcParams['figure.figsize'] = (9, 6)
plt.rcParams['font.size'] = 10
print("Librerías cargadas correctamente.")
```

## 1. Definición y Construcción de la Red A (`actividad_1_a.png`)

La imagen `actividad_1_a.png` representa un grafo dirigido con los siguientes costos/pesos en sus arcos:

* **Arcos:** $(1,2):5$, $(1,3):1$, $(3,2):2$, $(2,6):6$, $(6,2):7$, $(2,4):7$, $(2,5):1$, $(3,4):6$, $(4,3):7$, $(3,5):7$, $(4,6):4$, $(4,7):6$, $(5,4):3$, $(5,6):5$, $(5,7):9$, $(6,7):2$.

A continuación construimos tanto la versión **dirigida** (para Ruta Más Corta) como la **no dirigida** (para Árbol de Expansión Mínima).

```python
# 1. Definición del grafo dirigido G_a
G_a_dirigido = nx.DiGraph()

# Lista de conexiones del grafo con sus respectivos costos
aristas_a = [
    (1, 2, 5), (1, 3, 1), (3, 2, 2),
    (2, 6, 6), (6, 2, 7), (2, 4, 7), (2, 5, 1),
    (3, 4, 6), (4, 3, 7), (3, 5, 7),
    (4, 6, 4), (4, 7, 6), (5, 4, 3),
    (5, 6, 5), (5, 7, 9), (6, 7, 2)
]

# Agregar aristas con atributo 'weight'
for u, v, w in aristas_a:
    G_a_dirigido.add_edge(u, v, weight=w)

# Posiciones fijas para los nodos para que coincida con la imagen de la tarea
pos_a = {
    1: (0, 0),
    2: (1, 1),
    3: (1, -1),
    4: (2, 0),
    5: (2, -1.2),
    6: (3, 1),
    7: (4, 0)
}

# Dibujar el grafo dirigido
plt.figure()
nx.draw_networkx_nodes(G_a_dirigido, pos_a, node_size=700, node_color='coral', edgecolors='black')
nx.draw_networkx_labels(G_a_dirigido, pos_a, font_weight='bold')
nx.draw_networkx_edges(G_a_dirigido, pos_a, arrowstyle='->', arrowsize=15, edge_color='gray', connectionstyle='arc3,rad=0.08')

# Etiquetas de pesos
labels_a = nx.get_edge_attributes(G_a_dirigido, 'weight')
nx.draw_networkx_edge_labels(G_a_dirigido, pos_a, edge_labels=labels_a, label_pos=0.3)

plt.title("Red A (Dirigida) - Representación Original", fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

## 2. Problema 1: Árbol de Expansión Mínima (MST)

### Formulación Matemática
Dado un grafo no dirigido y conexo $G = (V, E)$ con pesos $w_{e} \ge 0$ para cada arista $e \in E$, el objetivo es encontrar un subgrafo $T = (V, E_T)$ tal que:
1. $T$ sea un árbol (conexo y sin ciclos).
2. $T$ contenga todos los vértices $|V|$.
3. Se minimice el costo total de las aristas:
$$\min \sum_{e \in E_T} w_e$$

**Nota:** Por requerimiento de la tarea, se toma la **versión no dirigida** de la Red A.

```python
# Convertir la Red A a su versión no dirigida
G_a_nodirigido = G_a_dirigido.to_undirected()

# Calculamos el MST usando la función integrada de NetworkX
mst_tree = nx.minimum_spanning_tree(G_a_nodirigido, algorithm='kruskal', weight='weight')

# Obtener aristas del MST y calcular peso total
mst_edges = list(mst_tree.edges(data=True))
costo_total_mst = sum(data['weight'] for u, v, data in mst_edges)

print("="*50)
print(" RESULTADO DEL ÁRBOL DE EXPANSIÓN MÍNIMA (MST)")
print("="*50)
print("Aristas seleccionadas en el MST:")
for u, v, data in mst_edges:
    print(f"  • Arista ({u} - {v}) con peso: {data['weight']}")
print(f"\n---> COSTO TOTAL MÍNIMO DEL MST = {costo_total_mst} <---")
print("="*50)

# Dibujar la solución destacando las aristas del MST
plt.figure()
nx.draw_networkx_nodes(G_a_nodirigido, pos_a, node_size=700, node_color='khaki', edgecolors='black')
nx.draw_networkx_labels(G_a_nodirigido, pos_a, font_weight='bold')

# Aristas no seleccionadas (gris claro) y seleccionadas (rojo grueso)
edge_colors = ['red' if G_a_nodirigido.has_edge(u, v) and mst_tree.has_edge(u, v) else 'lightgray'
               for u, v in G_a_nodirigido.edges()]
edge_widths = [3.0 if mst_tree.has_edge(u, v) else 1.0 for u, v in G_a_nodirigido.edges()]

nx.draw_networkx_edges(G_a_nodirigido, pos_a, edge_color=edge_colors, width=edge_widths)
nx.draw_networkx_edge_labels(G_a_nodirigido, pos_a, edge_labels=labels_a)

plt.title(f"Solución MST (Costo Total: {costo_total_mst})", fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

## 3. Problema 2: Ruta Más Corta (Shortest Path)

### Formulación Matemática
Dado un grafo dirigido $G = (V, E)$, un nodo origen $s \in V$ y un nodo destino $t \in V$, se busca determinar el camino $P_{s \to t}$ que minimice la suma de los pesos de los arcos transitados:
$$\min \sum_{(i,j) \in P_{s \to t}} w_{ij}$$

Sujeto a las restricciones de conservación de flujo:
$$\sum_{j} x_{ij} - \sum_{j} x_{ji} = \begin{cases} 1 & \text{si } i = s \\ -1 & \text{si } i = t \\ 0 & \text{en otro caso} \end{cases}$$

Para este ejercicio, consideraremos la ruta óptima desde el nodo origen **1** hasta el nodo destino **7**.

```python
origen = 1
destino = 7

# Obtenemos el camino mínimo y la distancia total del nodo 1 al 7
distancia_corta = nx.dijkstra_path_length(G_a_dirigido, source=origen, target=destino, weight='weight')
ruta_corta = nx.dijkstra_path(G_a_dirigido, source=origen, target=destino, weight='weight')

# Identificar los arcos que forman parte de la ruta óptima
arcos_ruta = list(zip(ruta_corta[:-1], ruta_corta[1:]))

print("="*50)
print(" RESULTADO DE LA RUTA MÁS CORTA")
print("="*50)
print(f"Origen: Nodo {origen}")
print(f"Destino: Nodo {destino}")
print(f"Ruta óptima trazada: {' -> '.join(map(str, ruta_corta))}")
print(f"\n---> DISTANCIA / COSTO TOTAL = {distancia_corta} <---")
print("="*50)

# Dibujar la solución destacando la ruta óptima
plt.figure()
nx.draw_networkx_nodes(G_a_dirigido, pos_a, node_size=700, node_color='mediumpurple', edgecolors='black')
nx.draw_networkx_labels(G_a_dirigido, pos_a, font_weight='bold')

# Resaltar arcos
edge_colors = ['purple' if (u, v) in arcos_ruta else 'lightgray' for u, v in G_a_dirigido.edges()]
edge_widths = [3.5 if (u, v) in arcos_ruta else 1.0 for u, v in G_a_dirigido.edges()]

nx.draw_networkx_edges(G_a_dirigido, pos_a, arrowstyle='->', arrowsize=15,
                       edge_color=edge_colors, width=edge_widths, connectionstyle='arc3,rad=0.08')
nx.draw_networkx_edge_labels(G_a_dirigido, pos_a, edge_labels=labels_a, label_pos=0.3)

plt.title(f"Ruta Más Corta ({origen} $\\to$ {destino}) - Costo: {distancia_corta}", fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

## 4. Problema 3: Flujo Máximo (Maximum Flow)

### Formulación Matemática
Dada una red dirigida $G = (V, E)$ donde cada arco $(i,j)$ posee una capacidad máxima $c_{ij} \ge 0$, se requiere encontrar la cantidad máxima de flujo que puede enviarse desde un nodo fuente $s$ hasta un nodo sumidero $t$:
$$\max v$$

**Sujeto a:**
1. **Conservación de flujo:**
$$\sum_{j} x_{ij} - \sum_{j} x_{ji} = \begin{cases} v & \text{si } i = s \\ -v & \text{si } i = t \\ 0 & \text{en otro caso} \end{cases}$$
2. **Restricción de capacidad:**
$$0 \le x_{ij} \le c_{ij}, \quad \forall (i,j) \in E$$

---
### Interpretación de la Red B (`actividad_1_b.png`)
En la imagen `actividad_1_b.png`, los pares de números en cada extremo corresponden a las capacidades dirigidas entre los pares de nodos[cite: 1]:
* Entre **1** y **3**: Capacidad $1 \to 3 = 14$, Capacidad $3 \to 1 = 0$[cite: 1]
* Entre **1** y **5**: Capacidad $1 \to 5 = 4$, Capacidad $5 \to 1 = 0$[cite: 1]
* Entre **1** y **2**: Capacidad $1 \to 2 = 8$, Capacidad $2 \to 1 = 0$[cite: 1]
* Entre **3** y **5**: Capacidad $3 \to 5 = 10$, Capacidad $5 \to 3 = 0$[cite: 1]
* Entre **3** y **4**: Capacidad $3 \to 4 = 9$, Capacidad $4 \to 3 = 7$[cite: 1]
* Entre **3** y **2**: Capacidad $3 \to 2 = 0$, Capacidad $2 \to 3 = 10$[cite: 1]
* Entre **2** y **5**: Capacidad $2 \to 5 = 6$, Capacidad $5 \to 2 = 0$[cite: 1]
* Entre **2** y **4**: Capacidad $2 \to 4 = 7$, Capacidad $4 \to 2 = 6$[cite: 1]
* Entre **4** y **5**: Capacidad $4 \to 5 = 5$, Capacidad $5 \to 4 = 0$[cite: 1]

Consideramos como Fuente el **Nodo 1** y Sumidero el **Nodo 5**[cite: 1].

```python
# Crear grafo dirigido para el problema de Flujo Máximo
G_b = nx.DiGraph()

# Construcción de la red B asignando las capacidades correspondientes
aristas_b = [
    (1, 3, 14), (1, 5, 4), (1, 2, 8),
    (3, 5, 10), (3, 4, 9), (4, 3, 7), (2, 3, 10),
    (2, 5, 6), (2, 4, 7), (4, 2, 6), (4, 5, 5)
]

for u, v, c in aristas_b:
    G_b.add_edge(u, v, capacity=c)

# Calcular el Flujo Máximo usando el algoritmo de Edmonds-Karp / Ford-Fulkerson
fuente = 1
sumidero = 5

valor_flujo_max, dict_flujo = nx.maximum_flow(G_b, fuente, sumidero, capacity='capacity')

print("="*50)
print(" RESULTADO DEL FLUJO MÁXIMO")
print("="*50)
print(f"Nodo Fuente (Origen): {fuente}")
print(f"Nodo Sumidero (Destino): {sumidero}")
print(f"\n---> VALOR DE FLUJO MÁXIMO TOTAL = {valor_flujo_max} <---")
print("="*50)
print("\nDistribución del Flujo por Arco (Flujo Enviado / Capacidad):")
for u in dict_flujo:
    for v in dict_flujo[u]:
        flujo = dict_flujo[u][v]
        cap = G_b[u][v]['capacity']
        if cap > 0 or flujo > 0:
            print(f"  • Arco ({u} -> {v}): {flujo} / {cap}")

# Posición geométrica similar al diagrama 1b
pos_b = {
    1: (0, 0),
    2: (1, -1),
    3: (1, 1),
    4: (2, -1),
    5: (2.5, 0)
}

# Dibujar el grafo y el flujo obtenido
plt.figure()
nx.draw_networkx_nodes(G_b, pos_b, node_size=800, node_color='sandybrown', edgecolors='black')
nx.draw_networkx_labels(G_b, pos_b, font_weight='bold')

# Crear etiquetas personalizadas que muestren "Flujo / Capacidad"
labels_flujo = {}
for u, v, d in G_b.edges(data=True):
    labels_flujo[(u, v)] = f"{dict_flujo[u][v]}/{d['capacity']}"

nx.draw_networkx_edges(G_b, pos_b, arrowstyle='->', arrowsize=15, edge_color='gray')
nx.draw_networkx_edge_labels(G_b, pos_b, edge_labels=labels_flujo, font_color='darkred')

plt.title(f"Flujo Máximo ({fuente} $\\to$ {sumidero}) - Total: {valor_flujo_max}", fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

## 5. Documentación de Funciones y Módulos de NetworkX Utilizados

| Función / Método | Módulo / Clase | Descripción Técnica y Parámetros |
| :--- | :--- | :--- |
| `nx.DiGraph()` | `networkx` | Crea un grafo dirigido que permite arcos unidireccionales o bidireccionales con pesos o capacidades. |
| `to_undirected()` | `nx.Graph` | Convierte un grafo dirigido en un grafo no dirigido sumando o conservando los atributos relevantes de las aristas. |
| `nx.minimum_spanning_tree()` | `networkx.algorithms.tree` | Calcula el Árbol de Expansión Mínima de un grafo no dirigido. Utiliza por defecto el algoritmo de Kruskal o Prim (`algorithm='kruskal'`). |
| `nx.dijkstra_path_length()` | `networkx.algorithms.shortest_paths` | Calcula la menor distancia entre un nodo origen (`source`) y un nodo destino (`target`) evaluando el peso indicado en el parámetro `weight`. |
| `nx.dijkstra_path()` | `networkx.algorithms.shortest_paths` | Retorna la secuencia ordenada de nodos que conforman la ruta más corta según el peso especificado. |
| `nx.maximum_flow()` | `networkx.algorithms.flow` | Determina el valor total del flujo máximo y el diccionario de flujos individuales entre una fuente y un sumidero basándose en el parámetro `capacity`. |
